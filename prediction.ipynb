```python
import pickle
import tensorflow as tf
from tensorflow.keras.models import load_model
import pandas as pd
import numpy as np
```

```python
model=load_model('model.h5')
```

loading scaler and encoders

```python
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender=pickle.load(file)

with open('one_hot_encoder_geo.pkl','rb') as file:
    one_hot_encoder_geo=pickle.load(file)  


with open('scaler.pkl','rb') as file:
    scaler =pickle.load(file)        
```

```python
input_data={
 'CreditScore':619,
 "Geography":"Spain",
 'Gender':"Male",
 'Age':35,
 'Tenure':3,
 'Balance':83807.86,
 'NumOfProducts':2,
 'HasCrCard':0,
 'IsActiveMember':1,
 'EstimatedSalary':112542.58,
}
input_df = pd.DataFrame([input_data])
```

```python
input_df
```

```python
geo_encoder=one_hot_encoder_geo.transform(input_df[['Geography']])
```

```python
geo_encoded_df=pd.DataFrame(geo_encoder.toarray(),columns=one_hot_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df=input_df.drop('Geography',axis=1)
```

```python
input_data=pd.concat([input_df,geo_encoded_df],axis=1)
input_data
```

Converting Gender Column

```python
input_data['Gender']=label_encoder_gender.transform(input_data['Gender'])
```

```python
input_data
```

### Scaling the data

```python
scaled_data=scaler.transform(input_data)
scaled_data
```

PREDICTION FOR NEW DATA

```python
prediction=model.predict(scaled_data)
prediction
```

```python
if prediction>0.5:
    print("The person will exited")
else:
    print("It will not exited")    

```
